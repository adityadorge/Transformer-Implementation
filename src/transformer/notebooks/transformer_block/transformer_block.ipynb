### **Root Mean Square Normalization (RMSNorm)**

$$\text{RMSNorm}(x) = \gamma \odot \frac{x}{\text{RMS}(x)}$$

#### **Variables & Definitions**

* **$x \in \mathbb{R}^d$:** The input vector (a single token's representation of dimension $d$).
* **$\text{RMS}(x) = \sqrt{\frac{1}{d} \sum_{i=1}^{d} x_i^2 + \epsilon}$:** The root mean square of the input vector, where $\epsilon$ is a small constant added for numerical stability (prevents division by zero).
* **$\gamma \in \mathbb{R}^d$:** Learnable scale parameters (typically initialized to $1$).
* **$\odot$:** Element-wise multiplication (Hadamard product).
